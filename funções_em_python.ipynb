**variáveis** **locais**

criadas dentro da função e só podem ser acessadas dentro da função em execução definida

```python
def minha_funcao():
  x = 10
  print(f"variável local: {x}")

minha_funcao()
```

**variáveis globais**

criadas fora de todas as funções, podendo ser acessadas em qualquer parte do código, incluindo dentro das funções; podem ser usadas em qualquer parte do seu programa

```python
y = 4

def outra_funcao():
  print(f"variável global: {y}")

outra_funcao()
print(f"variável global fora da função: {y}")
```

```python
def calcular_pagamento(qtd_horas, valor_hora):
  horas = float(qtd_horas)
  taxa = float(valor_hora)
  if horas <= 40:
    salario = horas * taxa
  else:
    salario = horas * taxa
  return salario

horas = input('Digite as horas: ')
taxa = input('Digite a taxa: ')
total_salario = calcular_pagamento(horas, taxa)
print(f'O valor de seus rendimentos é R$ {total_salario:.2f}')
```
